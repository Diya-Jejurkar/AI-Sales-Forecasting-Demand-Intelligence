```python
import pandas as pd
import numpy as np

df = pd.read_csv("AI_Sales_Forecasting_10000_Rows.csv")

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print("Shape:", df.shape)
print("Date range:", df["Order_Date"].min(), "to", df["Order_Date"].max())
```

```python
# Create daily sales data

daily = (
    df.groupby("Order_Date")
    .agg(
        Revenue=("Revenue", "sum"),
        Quantity=("Quantity", "sum"),
        Orders=("Order_ID", "nunique")
    )
    .reset_index()
)

daily["Order_Date"] = pd.to_datetime(daily["Order_Date"])

print("Daily observations:", len(daily))
print(daily.head())
```

```python
# Make sure both Order_Date columns have the same datetime type

daily["Order_Date"] = pd.to_datetime(daily["Order_Date"])

calendar = pd.DataFrame({
    "Order_Date": pd.date_range(
        start="2023-01-01",
        end="2025-12-31",
        freq="D"
    )
})

daily = calendar.merge(
    daily,
    on="Order_Date",
    how="left"
)

# Fill days with no sales
daily["Revenue"] = daily["Revenue"].fillna(0)
daily["Quantity"] = daily["Quantity"].fillna(0)
daily["Orders"] = daily["Orders"].fillna(0)

print("Daily observations after calendar:", len(daily))
print(daily.head())
print(daily.tail())
```

```python
# Create complete calendar from 2023 to 2025

calendar = pd.DataFrame({
    "Order_Date": pd.date_range(
        start="2023-01-01",
        end="2025-12-31",
        freq="D"
    )
})

daily = calendar.merge(
    daily,
    on="Order_Date",
    how="left"
)

daily["Revenue"] = daily["Revenue"].fillna(0)
daily["Quantity"] = daily["Quantity"].fillna(0)
daily["Orders"] = daily["Orders"].fillna(0)

print("Daily observations after calendar:", len(daily))
print(daily.head())
print(daily.tail())
```

```python
# Create forecasting features

daily["Day"] = daily["Order_Date"].dt.day
daily["Month"] = daily["Order_Date"].dt.month
daily["Year"] = daily["Order_Date"].dt.year
daily["DayOfWeek"] = daily["Order_Date"].dt.dayofweek
daily["Quarter"] = daily["Order_Date"].dt.quarter
daily["DayOfYear"] = daily["Order_Date"].dt.dayofyear
daily["WeekOfYear"] = daily["Order_Date"].dt.isocalendar().week.astype(int)

daily["IsWeekend"] = (
    daily["DayOfWeek"] >= 5
).astype(int)

daily["Trend"] = np.arange(len(daily))

# Lag features
daily["Lag_1"] = daily["Revenue"].shift(1)
daily["Lag_2"] = daily["Revenue"].shift(2)
daily["Lag_3"] = daily["Revenue"].shift(3)
daily["Lag_7"] = daily["Revenue"].shift(7)
daily["Lag_14"] = daily["Revenue"].shift(14)
daily["Lag_21"] = daily["Revenue"].shift(21)
daily["Lag_28"] = daily["Revenue"].shift(28)
daily["Lag_30"] = daily["Revenue"].shift(30)

# Rolling features
daily["Rolling_7"] = (
    daily["Revenue"]
    .shift(1)
    .rolling(7)
    .mean()
)

daily["Rolling_14"] = (
    daily["Revenue"]
    .shift(1)
    .rolling(14)
    .mean()
)

daily["Rolling_30"] = (
    daily["Revenue"]
    .shift(1)
    .rolling(30)
    .mean()
)

daily["RollingStd_7"] = (
    daily["Revenue"]
    .shift(1)
    .rolling(7)
    .std()
)

daily["RollingStd_30"] = (
    daily["Revenue"]
    .shift(1)
    .rolling(30)
    .std()
)

print("Features created successfully.")
print("Shape:", daily.shape)
print(daily.head())
```

```python
# Remove rows with missing lag/rolling values

daily_model = daily.dropna().reset_index(drop=True)

print("Rows available for modeling:", len(daily_model))
```

```python
# Define final model features

features = [
    "Day",
    "Month",
    "Year",
    "DayOfWeek",
    "Quarter",
    "DayOfYear",
    "WeekOfYear",
    "IsWeekend",
    "Trend",
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Lag_7",
    "Lag_14",
    "Lag_21",
    "Lag_28",
    "Lag_30",
    "Rolling_7",
    "Rolling_14",
    "Rolling_30",
    "RollingStd_7",
    "RollingStd_30"
]

X = daily_model[features]
y = daily_model["Revenue"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

final_model = RandomForestRegressor(
    n_estimators=500,
    max_features="sqrt",
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

final_model.fit(X, y)

print("Final Random Forest model trained successfully.")
```

```python
# Generate future dates for Jan-Jun 2026

future_dates = pd.date_range(
    start="2026-01-01",
    end="2026-06-30",
    freq="D"
)

# Keep historical revenue
history = daily_model[
    ["Order_Date", "Revenue"]
].copy()

# Create future dataframe
future = pd.DataFrame({
    "Order_Date": future_dates,
    "Revenue": np.nan
})

# Combine historical + future
forecast_data = pd.concat(
    [history, future],
    ignore_index=True
)

print("Future dates:", len(future_dates))
print("Forecast period:",
      future_dates.min(),
      "to",
      future_dates.max())
```

```python
# FIXED: Recursive Daily Revenue Forecast

# Recreate forecast_data
forecast_data = pd.concat(
    [
        daily_model[["Order_Date", "Revenue"]].copy(),
        future[["Order_Date", "Revenue"]].copy()
    ],
    ignore_index=True
)

# Make sure date is datetime
forecast_data["Order_Date"] = pd.to_datetime(
    forecast_data["Order_Date"]
)

# Calendar features
forecast_data["Day"] = forecast_data["Order_Date"].dt.day
forecast_data["Month"] = forecast_data["Order_Date"].dt.month
forecast_data["Year"] = forecast_data["Order_Date"].dt.year
forecast_data["DayOfWeek"] = forecast_data["Order_Date"].dt.dayofweek
forecast_data["Quarter"] = forecast_data["Order_Date"].dt.quarter
forecast_data["DayOfYear"] = forecast_data["Order_Date"].dt.dayofyear
forecast_data["WeekOfYear"] = (
    forecast_data["Order_Date"]
    .dt.isocalendar()
    .week
    .astype(int)
)

forecast_data["IsWeekend"] = (
    forecast_data["DayOfWeek"] >= 5
).astype(int)

forecast_data["Trend"] = np.arange(len(forecast_data))


# Forecast only future rows
future_start = len(daily_model)

for i in range(future_start, len(forecast_data)):

    # Lag features
    forecast_data["Lag_1"] = forecast_data["Revenue"].shift(1)
    forecast_data["Lag_2"] = forecast_data["Revenue"].shift(2)
    forecast_data["Lag_3"] = forecast_data["Revenue"].shift(3)
    forecast_data["Lag_7"] = forecast_data["Revenue"].shift(7)
    forecast_data["Lag_14"] = forecast_data["Revenue"].shift(14)
    forecast_data["Lag_21"] = forecast_data["Revenue"].shift(21)
    forecast_data["Lag_28"] = forecast_data["Revenue"].shift(28)
    forecast_data["Lag_30"] = forecast_data["Revenue"].shift(30)

    # Rolling features
    forecast_data["Rolling_7"] = (
        forecast_data["Revenue"]
        .shift(1)
        .rolling(7)
        .mean()
    )

    forecast_data["Rolling_14"] = (
        forecast_data["Revenue"]
        .shift(1)
        .rolling(14)
        .mean()
    )

    forecast_data["Rolling_30"] = (
        forecast_data["Revenue"]
        .shift(1)
        .rolling(30)
        .mean()
    )

    forecast_data["RollingStd_7"] = (
        forecast_data["Revenue"]
        .shift(1)
        .rolling(7)
        .std()
    )

    forecast_data["RollingStd_30"] = (
        forecast_data["Revenue"]
        .shift(1)
        .rolling(30)
        .std()
    )

    # Get features for current date
    row = forecast_data.loc[i, features]

    # Predict revenue
    prediction = final_model.predict(
        pd.DataFrame(
            [row],
            columns=features
        )
    )[0]

    # Prevent negative revenue
    forecast_data.loc[i, "Revenue"] = max(0, prediction)


print("Daily forecasting completed successfully.")
print()
print("Forecast period:")
print(
    forecast_data.loc[future_start, "Order_Date"],
    "to",
    forecast_data.loc[len(forecast_data) - 1, "Order_Date"]
)
```
